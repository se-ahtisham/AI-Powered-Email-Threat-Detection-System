# **NLP PREPROCESSING**

````python
lemmatizer = WordNetLemmatizer()

# These are usually removed (is, are, am, ... etc)from text because they don’t add strong meaning.
# But it will NOT remove these important words: not, never, urgent etc...

stop_words = set(stopwords.words("english")) - {"not", "no", "never", "urgent"}

def preprocess(text):
    text = text.lower()                                         # convert text to lowercase
    text = re.sub(r"https?://\S+|www\.\S+", "URL", text)        # replace URLs with URL token
    text = re.sub(r"\$[\d,]+", "MONEY", text)                   # replace money values with MONEY token
    text = re.sub(r"\b\d+\b", "NUMBER", text)                   # replace numbers with NUMBER token
    text = text.translate(str.maketrans("", "", string.punctuation))  # remove punctuation
    text = re.sub(r"\s+", " ", text).strip()                     # remove extra spaces
    tokens = word_tokenize(text)                                 # split text into words
    tokens = [lemmatizer.lemmatize(t) for t in tokens if t not in stop_words and len(t) > 2]  # remove stopwords + lemmatize
    return " ".join(tokens)                                    # join tokens back into cleaned text

# preprocess all emails
X = [preprocess(t) for t in X]
````

## TF-IDF

````python
tfidf = TfidfVectorizer(
    max_features=1000,
    ngram_range=(1, 2)
)

X_vec = tfidf.fit_transform(X)
````

Complement Naive Bayes:  A probability-based text classifier

**How it works:**
- Uses word frequency (TF-IDF features)
- Calculates the probability of each class (Phishing / Spam / Safe)

**Why it is used:**
- ✔ Very fast
- ✔ Works well on text data
- ✔ A great baseline model for NLP tasks

---

Logistic Regression: A linear classification model

**How it works:**
- Finds the relationship between words and labels
- Outputs a probability score for each class

**Parameters:**
- `max_iter=1000` → ensures the model converges properly
- `random_state=42` → makes results reproducible

**Why it is used:**
- ✔ Simple yet powerful
- ✔ Works very well with TF-IDF features
- ✔ Produces stable probability outputs

---
Support Vector Machine (SVM): A classifier that finds the best decision boundary between classes

**Why CalibratedClassifierCV is used:**
- `LinearSVC` does not output probabilities directly
- Calibration wraps it to convert outputs into probability scores

**Why it is used:**
- ✔ Very high accuracy
- ✔ Excellent on high-dimensional text data
- ✔ Captures complex patterns effectively

---

Ensemble Model — Voting Classifier: An ensemble that combines all three models

**How it works:**
```
NB  → probability
LR  → probability   →  Average  →  Final Predicted Label
SVM → probability
```

**Why soft voting:**
- ✔ Uses probabilities instead of hard class labels
- ✔ More accurate than any single model alone
- ✔ Produces more stable and reliable predictions


## 🎯 Summary

| Model | Role |
|-------|------|
| Naive Bayes | Fast probability-based classifier |
| Logistic Regression | Linear pattern learning model |
| SVM | High-accuracy boundary classifier |
| Voting Classifier | Combines all models for best result |

---

````python
nb = ComplementNB()

lr = LogisticRegression(
    max_iter=1000,
    random_state=42
)

svm = CalibratedClassifierCV(
    LinearSVC()
)

# ENSEMBLE MODEL

model = VotingClassifier(
    estimators=[
        ("nb", nb),
        ("lr", lr),
        ("svm", svm)
    ],
    voting="soft"
)

model.fit(X_vec, y)
````

## PREDICTION

````python
def predict_email(text):
    cleaned = preprocess(text)
    vec = tfidf.transform([cleaned])
    pred = model.predict(vec)[0]
    prob = np.max(model.predict_proba(vec))

    labels = {0: "Phishing", 1: "Spam", 2: "Safe"}

    print("Result:", labels[pred])
    print("Confidence:", round(prob * 100, 2), "%")
````

# **Import Pipeline**  
pipeline() is a ready-made AI function.  

It simplifies:  
*   Tokenization
*   Model loading
*   Prediction  

````python
from transformers import pipeline
````

````python
classifier = pipeline(
    "text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)
````

````python
def predict(text):
    result = classifier(text[:512])[0]

    label = result["label"]
    score = result["score"]

    print("\nConfidence:", round(score * 100, 2), "%")

    if score < 0.6:
        print("Uncertain Email")
    elif "NEGATIVE" in label.upper():
        print("Possibly Phishing / Spam")
    else:
        print("Likely Safe Email")
````

````python
email = input("Enter email: ")
predict(email)
````
